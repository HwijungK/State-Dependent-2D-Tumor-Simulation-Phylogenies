```python
from pathlib import Path

# Support running from either eden_sim/ or the project directory.
if Path("eden_sim").is_dir():
    from eden_sim.sim import DATA_DIR, World, visualize_world
    from eden_sim.log_to_fasta import convert_fasta
else:
    from sim import DATA_DIR, World, visualize_world
    from log_to_fasta import convert_fasta

LOG_FILE = DATA_DIR / "log_eden.tsv"
FASTA_FILE = DATA_DIR / "log.fasta"
```

```python
from IPython.display import clear_output
from time import sleep

#reusable for diff inputs
world = World(size = 80, time_step_hours = 2, target_cells = 1000, growth_model = "unrestricted", log_file = LOG_FILE)
center = world.size // 2
world.seed_cell(x = center, y = center, death_rate = .1, birth_rate = .5)

for i in range(1000):
    world.update()
    clear_output(wait = True)

    # print("Updates completed:", world.step)
    # print("Simulated hours:", world.step * world.time_step_hours)
    # print("Living cells:", world.cell_count)
    # print()
    #
    # visualize_world(world)
    # print("\n")

    if world.cell_count == 0 or world.cell_count >= world.target_cells:
        break

    sleep(.1)

print("*" * 10)
visualize_world(world)
world.end_sim()
```
